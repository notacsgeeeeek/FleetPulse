```python
from pathlib import Path

RAW_DATA = Path("../data/raw")

print("Raw data location:", RAW_DATA.resolve())
print("\nFiles found:")

for file in RAW_DATA.iterdir():
    print(file.name)
```

```python
import h5py

file_path = RAW_DATA / "N-CMAPSS_DS01-005.h5"

with h5py.File(file_path, "r") as f:
    print("Top-level contents:")
    for key in f.keys():
        print("-", key)
```

```python
with h5py.File(file_path, "r") as f:
    print("Dataset shapes:\n")

    for key in f.keys():
        print(f"{key:10} -> {f[key].shape}")
```

```python
with h5py.File(file_path, "r") as f:
    for key in ["A_var", "T_var", "W_var", "X_s_var", "X_v_var"]:
        print(f"\n--- {key} ---")
        print(f[key][:])
```

```python
with h5py.File(file_path, "r") as f:
    print("Y_dev first 20 values:")
    print(f["Y_dev"][:20])

    print("\nY_test first 20 values:")
    print(f["Y_test"][:20])
```

```python
import numpy as np

with h5py.File(file_path, "r") as f:
    y_dev = f["Y_dev"][:].flatten()
    y_test = f["Y_test"][:].flatten()

print("Y_dev")
print("Minimum:", y_dev.min())
print("Maximum:", y_dev.max())
print("Unique values:", np.unique(y_dev)[:30])
print("Number of unique values:", len(np.unique(y_dev)))

print("\nY_test")
print("Minimum:", y_test.min())
print("Maximum:", y_test.max())
print("Unique values:", np.unique(y_test)[:30])
print("Number of unique values:", len(np.unique(y_test)))
```

```python
with h5py.File(file_path, "r") as f:
    A_dev = f["A_dev"][:]

print("First 20 rows of A_dev:")
print(A_dev[:20])

print("\nNumber of unique engines:", len(np.unique(A_dev[:, 0])))

print("\nFirst 20 engine IDs:")
print(np.unique(A_dev[:, 0])[:20])
```

```python
with h5py.File(file_path, "r") as f:
    A_dev = f["A_dev"][:]
    Y_dev = f["Y_dev"][:].flatten()

engine_1 = A_dev[:, 0] == 1

engine_1_data = A_dev[engine_1]
engine_1_y = Y_dev[engine_1]

print("Engine 1 rows:", len(engine_1_data))
print("First 10 cycles:", engine_1_data[:10, 1])
print("Last 10 cycles:", engine_1_data[-10:, 1])

print("\nFirst 10 Y values:", engine_1_y[:10])
print("Last 10 Y values:", engine_1_y[-10:])

print("\nEngine 1 final cycle:", engine_1_data[-1, 1])
print("Engine 1 final Y:", engine_1_y[-1])
```

```python
with h5py.File(file_path, "r") as f:
    A_dev = f["A_dev"][:]
    W_dev = f["W_dev"][:]
    X_s_dev = f["X_s_dev"][:]
    X_v_dev = f["X_v_dev"][:]
    Y_dev = f["Y_dev"][:].flatten()

mask = (A_dev[:, 0] == 1) & (A_dev[:, 1] == 1)

print("Rows in Engine 1, Cycle 1:", mask.sum())

print("\nA values:")
print(A_dev[mask][:5])

print("\nW values:")
print(W_dev[mask][:5])

print("\nX_s values:")
print(X_s_dev[mask][:5])

print("\nX_v values:")
print(X_v_dev[mask][:5])

print("\nY values:")
print(Y_dev[mask][:20])
```
